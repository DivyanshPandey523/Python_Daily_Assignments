```python
students = {"Rahul": 75, "Amit": 32, "Priya": 88, "Neha": 39, "Rohit": 55}
for name, marks in students.items():
    if marks >= 40:
        print(name, "Pass")
    else:
        print(name, "Fail")
```

```python
students={"Rahul":85,"Amit":35,"Priya":72,"Neha":48}
for name,marks in students.items():
    if marks>=80:
        print(name,"Excellent")
    elif marks>=60:
        print(name,"Good")
    elif marks>=40:
        print(name,"Average")
    else:
        print(name,"Fail")
```

```python
product = {"Laptop": 10, "Mouse": 3, "Keyboard": 0, "Monitor": 7, "Printer": 2}
for item, quantity in product.items():
    if quantity == 0:
        print(item, "Out of stock")
    elif quantity <= 5:
        print(item, "Low stock")
    else:
        print(item, "Available")
```

```python
# A simple nested if example
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Please bring your ID")
else:
    print("You are not old enough")
```
